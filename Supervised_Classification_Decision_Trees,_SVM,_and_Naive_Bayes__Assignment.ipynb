2. Why Information Gain Is Important

Information Gain helps the decision tree answer:

👉 Which feature should I split on first?
👉 Which feature leads to the purest child nodes?
👉 Which feature gives the maximum reduction in randomness?

The feature that produces the highest Information Gain is selected as the root node, and the process continues recursively for each child node.

3. How Information Gain Is Used in Decision Trees

Here is the step-by-step process:

Step 1: Compute Entropy of the Parent Node

Before splitting, calculate entropy of the entire dataset.

Step 2: Compute Entropy for Each Possible Split

For each attribute, compute the entropy of child nodes created after splitting.

Step 3: Calculate Weighted Entropy

Child node entropies are weighted by the proportion of samples in each node.

Step 4: Compute Information Gain for Each Attribute

Subtract weighted child entropy from the parent entropy.

Step 5: Select Attribute with Maximum Information Gain

This attribute becomes the splitting attribute.

Step 6: Repeat the Process Recursively

The tree continues to grow until:

All nodes are pure

Or no more information gain is possible

Or stopping criteria (like pre-pruning) is met

Question 3:What is Pre-Pruning in Decision Trees?

Ans : re-pruning means you stop the split of the node early. Pre-pruning stops the split if the split does not give an improvement. Pre-pruning is simple.
We stop growing the tree. The tree stays simple.

How Pre-Pruning Works
During tree construction the algorithm checks the conditions. After that the algorithm splits the node.

I see the common stopping conditions include:
I set samples, per node to make sure minimum samples per node does not split when there are a few samples. For example do not split if minimum samples per node is, than ten samples.

- Maximum depth reached
- Minimum information gain / impurity reduction is too small
- Maximum number of leaf nodes
- Minimum samples required to split

If the condition is met the node does not split any more. The node then becomes a leaf node.
Why We Use Pre-Pruning

Advantages :
- Prevents overfitting early
- Faster training time
- Produces simpler, more interpretable trees

Disadvantages:

-Risk of stopping too early (underfitting)
- Might miss potentially good splits deeper in the tree

Question 4:Write a Python program to train a Decision Tree Classifier using Gini

Ans:

```python
# Python Program: Decision Tree Classifier using Gini Impurity
# and printing feature importances

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# Step 1: Load Dataset
iris = load_iris()
X = iris.data
y = iris.target

# Step 2: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 3: Create Decision Tree using Gini Impurity
clf = DecisionTreeClassifier(criterion='gini', random_state=42)

# Step 4: Train the model
clf.fit(X_train, y_train)

# Step 5: Print accuracy
accuracy = clf.score(X_test, y_test)
print("Test Accuracy:", accuracy)

# Step 6: Print feature importances
print("\nFeature Importances:")
for name, importance in zip(iris.feature_names, clf.feature_importances_):
    print(f"{name}: {importance}")
```

Question 7: Write a Python program to train two SVM classifiers with Linear and RBF
kernels on the Wine dataset, then compare their accuracies.

Ans :

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Step 1: Load the Wine dataset
wine = load_wine()
X = wine.data
y = wine.target

# Step 2: Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 3: Scale the features (important for SVM)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Step 4: Train SVM with Linear Kernel
svm_linear = SVC(kernel='linear')
svm_linear.fit(X_train, y_train)
linear_accuracy = svm_linear.score(X_test, y_test)

# Step 5: Train SVM with RBF Kernel
svm_rbf = SVC(kernel='rbf')
svm_rbf.fit(X_train, y_train)
rbf_accuracy = svm_rbf.score(X_test, y_test)

# Step 6: Print accuracy comparison
print("Accuracy with Linear Kernel:", linear_accuracy)
print("Accuracy with RBF Kernel:", rbf_accuracy)
```

Question 8: What is the Naïve Bayes classifier, and why is it called "Naïve"?

Ans : The Naïve Bayes classifier is a straightforward, supervised machine learning algorithm built on Bayes’ Theorem. People use it all the time for things like spam filters, figuring out sentiment in text, sorting documents, and even diagnosing medical conditions. It’s quick, doesn’t need much computing power, and gets the job done.

Here’s how it works: For any data point, Naïve Bayes calculates how likely it is to belong to each possible class. Then it just picks the class with the highest probability.

At the heart of it all is Bayes’ Theorem, which helps update what we know about a class after seeing some features. The formula looks like this:

P(C|X) = [P(X|C) * P(C)] / P(X)

Where:
- P(C|X) is the chance that the data belongs to class C, given the features X (that’s the “posterior probability”).
- P(X|C) is the likelihood — how likely we are to see those features if the data is actually in class C.
- P(C) is the prior probability of class C, or basically, how common that class is before we look at the features.
- P(X) is the evidence, which tells us how likely we are to see those features in general.

The classifier runs this calculation for each class and picks the one with the highest score. That’s how it makes its call.

2. Why Do We Call It “Naïve”?

Naïve Bayes gets its name because it makes a big, pretty unrealistic assumption: it treats every feature as if it’s totally independent from the others, as long as you know the class label. Real data almost never works this way. Take email spam filters, for example. Words like “discount” and “offer” usually show up together and are obviously related, but Naïve Bayes just ignores that and pretends they’re independent. That’s the “naïve” part. Still, even with this oversimplified view of the world, Naïve Bayes often does a great job, especially with text data, where the independence thing isn’t as far off as you’d think.

3. Why Does Naïve Bayes Work So Well, Even Though It’s Naïve?

High-dimensional data like text actually acts more independent than you’d expect. On top of that, the way Naïve Bayes estimates probabilities helps smooth out the fact that features aren’t truly independent. You don’t need much training data to get good results, either. Plus, it’s lightning fast and can handle huge datasets without breaking a sweat.

4. Types of Naïve Bayes Classifiers

Gaussian Naïve Bayes – Works with continuous data that follows a normal distribution.

Multinomial Naïve Bayes – Used for things like text classification and word counts.

Bernoulli Naïve Bayes – Good for binary or true/false features.

Question 9: Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve
Bayes, and Bernoulli Naïve Bayes

Ans : Naïve Bayes is a group of probabilistic classifiers built on Bayes’ Theorem, and it assumes each feature acts independently from the others. You’ll see different types of Naïve Bayes models, each one tailored to fit different kinds of input data.

The three most popular versions are:

Gaussian Naïve Bayes (GNB)
Multinomial Naïve Bayes (MNB)
Bernoulli Naïve Bayes (BNB)

Let’s break them down:

1. Gaussian Naïve Bayes (GNB)
Use this when your features are continuous numbers—anything like height, weight, age, or temperature—and they tend to follow a normal (bell curve) distribution. GNB looks at each feature in each class, figures out the mean and variance, and assumes values come from a Gaussian distribution.

You’ll see GNB pop up in places like the classic iris dataset, medical data (think blood pressure or glucose readings), and any situation with continuous sensor data.

What’s nice? It handles small datasets well and feels right at home with continuous data.

2. Multinomial Naïve Bayes (MNB)
Pick this one for count-based features, especially with text. If you’re counting how many times a word shows up in a document or tracking token frequency, MNB is your friend. It expects your data to follow a multinomial distribution.

You’ll find MNB behind the scenes in text classification, spam filters, grouping news articles, and even sentiment analysis. It shines with bag-of-words and TF-IDF features, and it’s seriously fast for natural language processing tasks.

3. Bernoulli Naïve Bayes (BNB)
BNB is for binary features—so, features that show up as 0 or 1, true or false, yes or no. Instead of caring about how many times something appears, BNB just checks whether it’s there or not.

Examples? Whether a word pops up in a document, if someone clicks or skips, or anything with on/off states. BNB assumes each feature follows a Bernoulli distribution and pays attention to both the presence and absence of features.

You’ll use BNB for text classification with 0/1 indicators, some recommendation systems, and document classification where only word presence matters. It’s a solid choice when you’re working with small datasets and only care about whether something’s there—not how often.

Question 10: Breast Cancer Dataset
Write a Python program to train a Gaussian Naïve Bayes classifier on the Breast Cancer
dataset and evaluate accuracy.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Step 1: Load Breast Cancer dataset
data = load_breast_cancer()
X = data.data
y = data.target

# Step 2: Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 3: Create Gaussian Naive Bayes classifier
gnb = GaussianNB()

# Step 4: Train the model
gnb.fit(X_train, y_train)

# Step 5: Make predictions
y_pred = gnb.predict(X_test)

# Step 6: Evaluate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy of Gaussian Naïve Bayes on Breast Cancer Dataset:", accuracy)
```
